```python
import numpy as np
import matplotlib.pyplot as plt

# Task 3a: Total iterations
N = 100000

# Step 1: Generate random points
x = np.random.uniform(-1, 1, N)
y = np.random.uniform(-1, 1, N)

# Step 2: Check points inside circle
inside_circle = x**2 + y**2 <= 1

# Step 3: Estimate pi
pi_estimate = 4 * np.sum(inside_circle) / N
print(f"Estimated value of π: {pi_estimate}")

# Task 3b: Visualization
plt.figure(figsize=(6,6))
plt.scatter(x[inside_circle], y[inside_circle], color='blue', s=1, label='Inside Circle')
plt.scatter(x[~inside_circle], y[~inside_circle], color='red', s=1, label='Outside Circle')

# Draw square and circle
circle = plt.Circle((0,0), 1, color='black', fill=False, linewidth=2, label='Circle')
plt.gca().add_patch(circle)
plt.xlim(-1, 1)
plt.ylim(-1, 1)
plt.gca().set_aspect('equal')
plt.title(f"Monte Carlo Estimation of π\nEstimated π = {pi_estimate}")
plt.legend()
plt.show()
```
